```python
#Pilhas no Python

#Podemos usar a estrutura de lista para implementar uma Pilha
pilha = []

#Usando a função append para adicionar items na pilha
pilha.append("A")
pilha.append("B")
pilha.append("C")

#Imprimindo a pilha
print(f"Pilha: {pilha}")

#Usando o pop, removemos o último elemento (topo) da pilha
elemento_pilha = pilha.pop()

print(f"Elemento removido da pilha: {elemento_pilha}")
print(f"Pilha atualizada: {pilha}")
```

```python
#Filas no Python

#Criando a fila à partir da biblioteca collections
from collections import deque
fila = deque()

#Utilizando a função append para adicionar items na fila
fila.append("X") #Enfileirando "X"
fila.append("Y") #Enfileirando "Y"
fila.append("Z") #Enfileirando "Z"

#Imprimindo a fila em forma de lista
print("\nFila:", list(fila))

elemento_fila = fila.popleft() #Remove o primeiro elemento da fila

print("Elemento removido da fila:", elemento_fila)
print("Fila depois de remover o item:", list(fila))


```
