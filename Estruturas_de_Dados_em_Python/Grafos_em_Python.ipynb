```python
###Exemplo de Grafo###

class Grafo:
    def __init__(self):
        self.grafo = {} #lembrando que o grafo se comporta como um dict.

    #Definindo função adicionar_vertice
    def adicionar_vertice(self, vertice): 
        if vertice not in self.grafo:
            self.grafo[vertice] = []

    #definindo função adicionar_aresta, solicitando o vertice de origem, o vertice de destino e atribuindo o peso.
    def adicionar_aresta(self, origem, destino, peso = 1): 
        if origem not in self.grafo:
            self.adicionar_vertice(origem)
        if destino not in self.grafo:
            self.adicionar_vertice(destino)
        self.grafo[origem].append((destino, peso))

    #Criando função para representar o grafo em uma string.
    def mostrar_grafo(self):
        for vertice, arestas in self.grafo.items():
            print(f"{vertice}: {arestas}")


###Exemplos de execução###

grafo = Grafo()
grafo.adicionar_vertice("A")
grafo.adicionar_vertice("B")
grafo.adicionar_vertice("C")
grafo.adicionar_aresta("A", "B", 5)
grafo.adicionar_aresta("A", "C", 10)
grafo.adicionar_aresta("B", "C", 3)
grafo.mostrar_grafo()

```
