```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = int(preco)

    def printnp(self):
        return f"O produto se chama {self.nome} e custa {self.preco}."
    
    def aplicar_desconto(self, desconto):
        self.preco = int(self.preco * ((100 - desconto) / 100))

    def aplicar_aumento(self, acrescimo):
        self.preco = int(self.preco * ((100 + acrescimo) / 100))

    
videogame = Produto("Playstation", "2500")

videogame.aplicar_aumento(10)
videogame.printnp()

# videogame.aplicar_desconto(50)
# videogame.printnp()


class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_gb):
        super().__init__(nome, preco)
        self.tamanho_gb = tamanho_gb



```
