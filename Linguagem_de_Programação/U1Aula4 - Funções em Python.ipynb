```python
#Função len()

#Cria uma lista de numeros
numeros = [1, 2, 3, 4, 5]


#Usa a função len() para calcular o comprimento da lista
comprimento = len(numeros)

#Imprime o comprimento da lista
print("O comprimento da lista é:", comprimento)

```
