```python
#Explorando Classes

#Definindo uma classe chamada Pessoa.

class Pessoa:
    #O método __init__ é um construtor, chamando quando um objeto da classe é criado.
    #Ele inicializa os atributos da classe.
    def __init__(self, nome , idade, genero):
        # self é uma convenção em Python quue se refere à própria instância da classe.
        #Os parâmetros nome, idade e genero são passados durante a criação do objeto.
        #Eles são usados para inicializar os atribuitos da instância.
        self.nome = nome #Atribui o valor de nome ao atributo nome da instância.
        self.idade = idade #Atribui o valor de idade ao atributo idade da instância.
        self.genero = genero #Atribui o valor de genero ao atributo genero da instância.

        #o método cumprimentar retorna um alô com o nome da pessoa

    def cumprimentar(self):
        return f"Olá, tenha um bom dia {self.nome}"
    
        #Adicionando também o método aniversário, para que aumente a idade da pessoa em 1.

    def aniversario(self):
        self.idade += 1

        
```

```python
#Criando uma instância da classe Pessoa com os valores.

pessoa1 = Pessoa("Raphael", 35, "Masculino")

#Chama o método cumprimentar na sua instância pessoal e imprime a saudação.

print(pessoa1.cumprimentar())

#Acessando o atributo idade da instância pessoal e imprime sua idade

print(f"Idade: {pessoa1.idade}")

pessoa1.aniversario()

print(f"Nova idade: {pessoa1.idade}")
```

```python
#Explorando heranças 

class Animal:
    def __init__(self, nome):
        self.nome = nome
    def fazer_barulho(self):
        pass

class Cachorro(Animal):
    def fazer_barulho(self):
        return "Latir"

class Gato(Animal):
    def fazer_barulho(self):
        return "Miar"
    
    
```

```python
rex = Cachorro("Rex")

whiskers = Gato("Whiskers")

print(f"{rex.nome} faz: {rex.fazer_barulho()}")

print(f"{whiskers.nome} faz {whiskers.fazer_barulho()}")
```
