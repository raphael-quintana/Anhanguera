```python
#Explorando leitura de dados com pandas, lendo de uma API

#Importando pandas
import pandas as pd

#Criando a variável df_selic para armazenar o dataframe, e solicitando a leitura
#dos dados por API, com o método JSON.

df_selic = pd.read_json("https://api.bcb.gov.br/dados/serie/bcdata.sgs.11/dados?formato=json&dataInicial=11/04/2025")
#Tive que adicionar data inicial por solicitação da API

#Imprimindo o DF e o info()
print(df_selic)
print(df_selic.info())


```

```python
#Para verificar duplicidadede linhas (MUITO IMPORTANTE), podemos utilizar a
#função drop_duplicates()

df_selic.drop_duplicates(keep="last", inplace=True)
#Keep = last para manter o último registro
#Através do parâmetro inplace=True, faz com que a transformaçao seja salva no DF.

#No caso, não existem linhas duplicadas

print(df_selic)
```

```python
#Adicionando colunas no DF

#Importando datetime

from datetime import datetime as dt

#Salvando a data de hoje na variável data_extracao
data_extracao = dt.today()

#criando colunas no DF
df_selic["data_extracao"] = data_extracao
df_selic["responsavel"] = "Raphael"

#Imprimindo info e cabeca da DF.
print(df_selic.info())
df_selic.head()
```

```python
#Podemos localizar um dado individual no DF com a função loc e o indice entre
# chaves.
df_selic.loc[4]

#Também podemos fazer com mais de um, porém assim já se imprime a tabela
df_selic.loc[[0,22,75]]



```
