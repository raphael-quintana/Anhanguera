```python
numero = int(input("Digite um numero (ou 0 para sair): "))

while numero != 0:
    if numero % 2 == 0:
        print("O Numero e par.")
    else:
        print("O numero e impar")
    numero = int(input("Digite um numero (ou 0 para sair): "))
```
