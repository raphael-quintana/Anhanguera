```python
#explorando visualização de dados

#Importando bibliotecas
import matplotlib.pyplot as plt
import random


#Aqui criamos duas listas aleatorias utilizando o random, utilizamos o range
#para que seja limitado até 100, e o K indica a quantidade de elementos na lista.
dados1 = random.sample(range(100), k=20)
dados2 = random.sample(range(100), k=20)

print(dados1)
print(dados2)

plt.plot(dados1, dados2)
```

```python
#Visualização de Dados com Pandas

#Importando a biblioteca
import pandas as pd

#Criando um dicionario de produtos vendidos
dados = {
    "Produto": ["A", "B", "C"],
    "qtd_vendida": [33, 50, 45]
}

#Criando o dataframe com o dicionário.
df = pd.DataFrame(dados)

#Criando o gráfico com o método plot()
df.plot(x="Produto", y="qtd_vendida", kind="bar")
```

```python
df.plot(x="Produto", y="qtd_vendida", kind="pie")
```

```python
df.plot(x="Produto", y="qtd_vendida", kind="line")
```

```python
#Explorando a visualização de dados com Seaborn (DATASET pronto do seaborn)

#Importando bibliotecas
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid") #opcoes: darkgrid, whitegrid, dark, white, ticks

df_tips = sns.load_dataset("tips") #dataset disponivel no github do Seaborn

print(df_tips)

```

```python
#usando subplots do matplotlib para fazer mais graficos de uma só vez.
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

sns.barplot(data=df_tips, x="sex", y="total_bill", ax=ax[0]) #Gasto médio por sexo.

sns.barplot(data=df_tips, x="sex", y="total_bill", ax=ax[1], estimator=sum) #Soma total por sexo

sns.barplot(data=df_tips, x="sex", y="total_bill", ax=ax[2], estimator=len) #Soma de ocorrencias por sexo

```
