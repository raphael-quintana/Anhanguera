```python
#explorando visualização de dados

#Importando bibliotecas
import matplotlib.pyplot as plt
import random


#Aqui criamos duas listas aleatorias utilizando o random, utilizamos o range
#para que seja limitado até 100, e o K indica a quantidade de elementos na lista.
dados1 = random.sample(range(100), k=20)
dados2 = random.sample(range(100), k=20)

print(dados1)
print(dados2)

plt.plot(dados1, dados2)
```
