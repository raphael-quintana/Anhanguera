```python
#Explorando os sets (conjuntos)

meu_conjunto = set()

meu_conjunto


```

```python
#Adicionando elementos ao conjunto

meu_conjunto.add(10)
meu_conjunto.add(20)
meu_conjunto.add(30)

#Imprimindo o conjunto

print("Conjunto após adicionar elementos:", meu_conjunto)
```

```python
#Verificando se um elemento está no conjunto

elemento = 20
if elemento in meu_conjunto:
    print(f"{elemento} está no conjunto.")
else:
    print(f"{elemento} não está no conjunto.")

#Removendo um elemento do conjunto
meu_conjunto.remove(20)


#Imprimindo o conjunto atualizado

print("Conjunto após remover o elemento 20:", meu_conjunto)

```

```python
#Explorando Dicionários

#Criando um dicionário vazio, seguido de atribuição de chaves e valores

dici_1 = {}
dici_1["nome"] = "Maria"
dici_1["idade"] = 25

#Criando um dicionário com pares de chave:valor

dici_2 = {"nome": "Maria", "idade": 25}

#Criando um dicionário com uma lista de tuplas representando pares chave:valor

dici_3 = dict([("nome", "Maria"), ("idade", 25)])

#Criando um dicionário usando a função built-in chamada zip(), e duas listas,
#uma para chaves, e outra para os valores.

dici_4 = dict(zip(["nome", "idade"], ["Maria", 25]))

#Testando se todas as construções resultam em objetos iguais

print(dici_1 == dici_2 == dici_3 == dici_4)
print(dici_1)
```

```python
#Explorando a biblioteca NumPy.


#Importando

import numpy as np

#Criando um array NumPy de números inteiros

my_array = np.array([1, 2, 3, 4, 5])

#Imprimindo o array

print("Array original:")
print(my_array)

#Realizando operações matemáticas com o array

squared_array = my_array ** 2 #Elevando cada elemento ao quadrado.
sum_of_elements = np.sum(my_array) #Calcuando a soma de todos os elementos.

print("Array ao quadrado:")
print(squared_array)
print("Soma dos elementos:")
print(sum_of_elements)

#Acessando elementos do array pelo numero do índice em [].

element_at_index_2 = my_array[2]
print("\nElemento no índice 2:", element_at_index_2)

```
