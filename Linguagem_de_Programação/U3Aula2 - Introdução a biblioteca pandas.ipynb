```python
#Explorando Pandas

#Criando uma Series com Pandas

#Importando Pandas
import pandas as pd

#Criando uma lista de valores
exemplo1 = [10, 20, 30, 40, 50]

#Criando uma Series a partir da lista
series1 = pd.Series(data = exemplo1)

print(series1)
```

```python
#Podemos também criar uma Series à partir de um dicionário.

#Importando o pandas
import pandas as pd

#Criando um dicionário com pares chave-valor
exemplo2 = {"A": 100, "B": 200, "C": 300, "D": 400, "E": 500}

#Criando a Series à partir do dicionário.
series2 = pd.Series(data = exemplo2)

print(series2)
print(type(series2))
print(len(series2))
```

```python
#Leitura de dados estruturados com Pandas

#Importando o Pandas
import pandas as pd

#Adicionando a variável com a url, e solicitando a leitura com o método read
url = "https://www.fdic.gov/resources/resolutions/bank-failures/failed-bank-list/"
dfs = pd.read_html(url)


#ele nos retorna uma lista, com diversos dfs
print(type(dfs))
print(len(dfs))
```

```python

#Criando o DF bancos, com o primeiro DF da lista criada.
df_bancos = dfs[0]


#Repare que ele pegou os 25 primeiros bancos da página(25 linhas) e as 7 colunas.
print(df_bancos.shape)
```

```python
#Imprimindo os data types das colunas presentes no DF bancos.
print(df_bancos.dtypes)
```
